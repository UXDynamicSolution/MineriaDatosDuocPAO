# Clasificación de personas con if-else vs Árbol de Decisión (CLP)
Ejemplo educativo para minería de datos, usando edad y salario en pesos chilenos.

```python
# Datos de ejemplo: edad en años, salario en pesos chilenos
datos = [
    [15, 100000],
    [25, 450000],
    [65, 250000],
    [35, 900000],
    [70, 200000]
]
y = ["menor de edad", "adulto", "adulto mayor", "adulto", "adulto mayor"]
```

```python
# Clasificación con if-else
def clasificar_persona_if_else(edad, salario):
    if salario <= 400000:
        if edad <= 40:
            return "menor de edad"
        else:
            return "adulto mayor"
    else:
        return "adulto"

print("=== Clasificación con if-else ===")
for persona in datos:
    edad, salario = persona
    print(f"Edad: {edad}, Salario: ${salario:,} CLP => {clasificar_persona_if_else(edad, salario)}")
```

```python
# Clasificación con árbol de decisión
from sklearn.tree import DecisionTreeClassifier

modelo = DecisionTreeClassifier()
modelo.fit(datos, y)

print("\n=== Clasificación con árbol de decisión ===")
predicciones = modelo.predict(datos)
for persona, pred in zip(datos, predicciones):
    print(f"Edad: {persona[0]}, Salario: ${persona[1]:,} CLP => {pred}")
```
