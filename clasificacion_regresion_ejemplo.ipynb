```python
# === Importaciones necesarias ===
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LinearRegression
from sklearn.metrics import accuracy_score, r2_score
```

```python
# === Datos para CLASIFICACIÓN ===
datos_clasificacion = pd.DataFrame({
    'horas_estudio': [1, 2, 3, 4, 5, 6, 7, 8],
    'asistencias':   [2, 3, 4, 4, 5, 6, 6, 7],
    'aprueba':       [0, 0, 0, 0, 1, 1, 1, 1]
})

X_clf = datos_clasificacion[['horas_estudio', 'asistencias']]
y_clf = datos_clasificacion['aprueba']

# División en entrenamiento y prueba
Xc_train, Xc_test, yc_train, yc_test = train_test_split(X_clf, y_clf, test_size=0.3)

# Modelo de clasificación
modelo_clf = DecisionTreeClassifier()
modelo_clf.fit(Xc_train, yc_train)

# Predicción
yc_pred = modelo_clf.predict(Xc_test)
print("=== Clasificación ===")
print("Accuracy:", accuracy_score(yc_test, yc_pred))

# Probar un nuevo caso
nuevo_estudiante = [[5, 5]]
print("¿Aprueba con 5h estudio y 5 asistencias?:", modelo_clf.predict(nuevo_estudiante)[0])
```

```python
# === Datos para REGRESIÓN ===
datos_regresion = pd.DataFrame({
    'm2': [50, 60, 70, 80, 90, 100],
    'precio': [100, 120, 140, 160, 180, 200]
})

X_reg = datos_regresion[['m2']]
y_reg = datos_regresion['precio']

# División en entrenamiento y prueba
Xr_train, Xr_test, yr_train, yr_test = train_test_split(X_reg, y_reg, test_size=0.3)

# Modelo de regresión
modelo_reg = LinearRegression()
modelo_reg.fit(Xr_train, yr_train)

# Predicción
yr_pred = modelo_reg.predict(Xr_test)
print("
=== Regresión ===")
print("R² (score):", r2_score(yr_test, yr_pred))

# Predecir nuevo valor
m2_nuevo = [[75]]
print("Precio estimado para 75m2:", modelo_reg.predict(m2_nuevo)[0])
```

```python
# Visualización de regresión
plt.scatter(X_reg, y_reg, color='blue', label='Datos reales')
plt.plot(X_reg, modelo_reg.predict(X_reg), color='red', label='Modelo lineal')
plt.xlabel('m2 construidos')
plt.ylabel('Precio (UF)')
plt.title('Regresión lineal: Precio vs m2')
plt.legend()
plt.grid()
plt.show()
```
