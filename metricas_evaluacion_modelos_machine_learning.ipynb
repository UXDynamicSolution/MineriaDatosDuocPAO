# Métricas de Evaluación en Machine Learning

| Métrica | Tipo de Problema | ¿Qué responde? | Rango Ideal | ¿Cuándo priorizarla? |
|---|---|---|---|---|
| **Exactitud (Accuracy)** | Clasificación | ¿Qué porcentaje total de predicciones fueron correctas? | Cercano a 1 (100%) | Cuando las clases están balanceadas. Puede ser engañosa en datasets desbalanceados. |
| **Precisión (Precision)** | Clasificación | De todos los casos predichos como positivos, ¿cuántos realmente eran positivos? | Cercano a 1 | Cuando el costo de un Falso Positivo es alto (ej. detectar spam incorrectamente). |
| **Sensibilidad (Recall)** | Clasificación | De todos los positivos reales existentes, ¿cuántos detectó el modelo? | Cercano a 1 | Cuando el costo de un Falso Negativo es crítico (ej. diagnóstico médico). |
| **F1-Score** | Clasificación | ¿Cuál es el equilibrio entre Precision y Recall? | Cercano a 1 | En datasets desbalanceados donde importan tanto Falsos Positivos como Falsos Negativos. |
| **AUC-ROC** | Clasificación | ¿Qué tan bien separa el modelo las clases positivas y negativas independientemente del umbral? | 0.5 a 1.0 | Para comparar el desempeño general entre modelos de clasificación. |
| **R Cuadrado (R²)** | Regresión | ¿Qué proporción de la variabilidad del dato es explicada por el modelo? | Cercano a 1 | Para medir la capacidad explicativa global del modelo de regresión. |
| **MAE (Mean Absolute Error)** | Regresión | ¿Cuál es el error promedio absoluto de las predicciones? | Cercano a 0 | Cuando se desea interpretabilidad simple y los outliers no son críticos. |
| **RMSE (Root Mean Squared Error)** | Regresión | ¿Cuál es el error promedio penalizando más los errores grandes? | Cercano a 0 | Cuando los errores grandes deben castigarse fuertemente. |

---

# Interpretación General

## Clasificación

- **Accuracy** → rendimiento general.
- **Precision** → minimiza falsos positivos.
- **Recall** → minimiza falsos negativos.
- **F1-Score** → equilibrio entre Precision y Recall.
- **AUC-ROC** → capacidad discriminativa global del modelo.

## Regresión

- **R²** → calidad del ajuste.
- **MAE** → error promedio interpretable.
- **RMSE** → penaliza errores grandes.

---

# Ejemplo Práctico

## Diagnóstico Médico

- Si una enfermedad es grave:
  - Priorizar **Recall** para no dejar enfermos sin detectar.
- Si el tratamiento es riesgoso o caro:
  - Priorizar **Precision** para evitar falsos positivos.

## Predicción de Precios

- **MAE** → útil para conocer el error promedio real.
- **RMSE** → útil si grandes errores financieros son peligrosos.


# Rangos e Interpretación de Métricas en Machine Learning

## Métricas de Clasificación

| Métrica | Valor Mínimo | Significado del Mínimo | Valor Máximo | Significado del Máximo |
|---|---|---|---|---|
| **Accuracy (Exactitud)** | 0 | Ninguna predicción fue correcta. | 1 | Todas las predicciones fueron correctas. |
| **Precision** | 0 | Todos los positivos predichos fueron incorrectos (muchos falsos positivos). | 1 | Todos los positivos predichos eran realmente positivos. |
| **Recall (Sensibilidad)** | 0 | El modelo no detectó ningún positivo real. | 1 | El modelo detectó todos los positivos reales. |
| **F1-Score** | 0 | No existe equilibrio entre Precision y Recall; desempeño muy pobre. | 1 | Precision y Recall son perfectos. |
| **AUC-ROC** | 0 | El modelo clasifica completamente al revés de lo esperado. | 1 | Separación perfecta entre clases positivas y negativas. |

---

# Interpretación de AUC-ROC

| Rango | Interpretación |
|---|---|
| 0.5 | Modelo equivalente al azar |
| 0.6 – 0.7 | Rendimiento pobre |
| 0.7 – 0.8 | Rendimiento aceptable |
| 0.8 – 0.9 | Buen modelo |
| 0.9 – 1.0 | Excelente discriminación |

---

# Métricas de Regresión

| Métrica | Valor Mínimo | Significado del Mínimo | Valor Máximo | Significado del Máximo |
|---|---|---|---|---|
| **R² (Coeficiente de Determinación)** | Puede ser negativo | El modelo es peor que predecir usando la media. | 1 | El modelo explica perfectamente la variabilidad de los datos. |
| **MAE** | 0 | No existe error promedio en las predicciones. | ∞ | Error promedio extremadamente alto. |
| **RMSE** | 0 | No existe error cuadrático en las predicciones. | ∞ | Existen errores muy grandes en las predicciones. |

---

# Interpretación General de R²

| Rango | Interpretación |
|---|---|
| Menor que 0 | Modelo muy deficiente |
| 0 – 0.5 | Baja capacidad explicativa |
| 0.5 – 0.7 | Ajuste moderado |
| 0.7 – 0.9 | Buen ajuste |
| 0.9 – 1.0 | Excelente ajuste |

---

# Interpretación General de MAE y RMSE

## MAE

- Representa el error promedio absoluto.
- Se expresa en las mismas unidades del problema.
- Más fácil de interpretar.

### Ejemplo
Si el MAE = 2:
- En promedio, el modelo se equivoca en 2 unidades.

---

## RMSE

- Penaliza más los errores grandes.
- Sensible a outliers.
- Muy utilizado cuando grandes errores son peligrosos.

### Ejemplo
Si el RMSE = 10:
- Existen errores grandes que están aumentando considerablemente el promedio cuadrático.
