# Métricas evaluación modelos de machine learning (clase 2.1.1)

En minería de datos y aprendizaje automático, las métricas permiten evaluar la calidad, relevancia y desempeño de patrones, reglas de asociación y modelos predictivos.

---

# 1. Soporte (Support)

## ¿Qué mide?
Mide la frecuencia con la que aparece un patrón dentro del conjunto de datos.

## Fórmula

\[
Support(A) = \frac{\text{Número de transacciones que contienen } A}{\text{Total de transacciones}}
\]

## Interpretación
- Valor alto → El patrón aparece frecuentemente.
- Valor bajo → El patrón es raro.

## Rango
| Valor | Significado |
|---|---|
| 0 | El patrón nunca aparece |
| 1 | El patrón aparece en todas las transacciones |

## Ejemplo
Si 30 de 100 compras contienen pan:

\[
Support(Pan) = \frac{30}{100} = 0.3
\]

El soporte es 30%.

---

# 2. Confianza (Confidence)

## ¿Qué mide?
Mide la probabilidad de que ocurra un patrón B dado que ocurrió A.

## Fórmula

\[
Confidence(A \rightarrow B) = \frac{Support(A \cap B)}{Support(A)}
\]

## Interpretación
- Alta confianza → Fuerte relación entre A y B.
- Baja confianza → Asociación débil.

## Rango
| Valor | Significado |
|---|---|
| 0 | Nunca ocurre B cuando ocurre A |
| 1 | Siempre ocurre B cuando ocurre A |

## Ejemplo
Si 20 personas compran pan y leche, y 25 compran pan:

\[
Confidence(Pan \rightarrow Leche) = \frac{20}{25} = 0.8
\]

Existe un 80% de probabilidad de comprar leche si se compra pan.

---

# 3. Lift

## ¿Qué mide?
Evalúa qué tan fuerte es la asociación entre dos patrones comparado con si fueran independientes.

## Fórmula

\[
Lift(A \rightarrow B) = \frac{Confidence(A \rightarrow B)}{Support(B)}
\]

## Interpretación
| Valor | Significado |
|---|---|
| < 1 | Asociación negativa |
| = 1 | Independencia |
| > 1 | Asociación positiva |

## Ejemplo
Si:

- Confidence = 0.8
- Support(B) = 0.4

\[
Lift = \frac{0.8}{0.4} = 2
\]

La ocurrencia conjunta es 2 veces más probable de lo esperado.

---

# 4. Convicción (Conviction)

## ¿Qué mide?
Mide el grado de dependencia de una regla de asociación.

## Fórmula

\[
Conviction(A \rightarrow B) =
\frac{1 - Support(B)}
{1 - Confidence(A \rightarrow B)}
\]

## Interpretación
- Valor cercano a 1 → Baja dependencia.
- Valor alto → Fuerte dependencia.

## Rango
| Valor | Significado |
|---|---|
| 1 | Independencia |
| > 1 | Dependencia positiva |
| ∞ | Regla perfecta |

## Ejemplo
Una convicción de 3 indica que la regla es 3 veces más fuerte que una asociación aleatoria.

---

# 5. Jaccard Similarity

## ¿Qué mide?
Mide la similitud entre dos conjuntos.

## Fórmula

\[
J(A,B) =
\frac{|A \cap B|}
{|A \cup B|}
\]

## Interpretación
- Alto → Conjuntos similares.
- Bajo → Conjuntos distintos.

## Rango
| Valor | Significado |
|---|---|
| 0 | No comparten elementos |
| 1 | Son idénticos |

## Ejemplo
Si dos usuarios tienen 3 gustos en común y 5 diferentes:

\[
J = \frac{3}{8} = 0.375
\]

---

# 6. Kulczynski Measure

## ¿Qué mide?
Promedia la probabilidad de A dado B y de B dado A.

## Fórmula

\[
Kulc =
\frac{1}{2}
\left(
P(A|B) + P(B|A)
\right)
\]

## Interpretación
- Alto → Relación equilibrada y fuerte.
- Bajo → Relación débil.

## Rango
| Valor | Significado |
|---|---|
| 0 | Sin asociación |
| 1 | Asociación perfecta |

## Uso común
- Clasificación
- Reglas de asociación
- Datos desbalanceados

---

# 7. IR (Interest Rate)

## ¿Qué mide?
Evalúa el interés o relevancia de un patrón respecto al conjunto de datos.

## Interpretación
- Alto → Patrón interesante o poco común.
- Bajo → Patrón poco relevante.

## Uso común
- Descubrimiento de patrones
- Reglas de asociación

## Rango típico
No posee un rango universal fijo; depende de la implementación utilizada.

---

# 8. Índice Gini (Gini Index)

## ¿Qué mide?
Mide la impureza de un conjunto de datos.

## Fórmula

\[
Gini = 1 - \sum p_i^2
\]

Donde:

- \(p_i\) = probabilidad de cada clase.

## Interpretación
- Bajo → Datos más puros.
- Alto → Datos mezclados.

## Rango
| Valor | Significado |
|---|---|
| 0 | Nodo completamente puro |
| Cercano a 1 | Alta mezcla de clases |

## Ejemplo
Usado en:
- Árboles de decisión
- Random Forest

---

# 9. ROC-AUC

## ¿Qué mide?
Evalúa la capacidad de un modelo para distinguir clases positivas y negativas.

## Componentes
- ROC: Curva de verdaderos positivos vs falsos positivos.
- AUC: Área bajo la curva ROC.

## Interpretación
| Valor | Significado |
|---|---|
| 0.5 | Modelo aleatorio |
| 0.7 - 0.8 | Buen modelo |
| 0.8 - 0.9 | Muy buen modelo |
| > 0.9 | Excelente modelo |

## Rango
| Valor | Significado |
|---|---|
| 0 | Clasificación incorrecta total |
| 1 | Clasificación perfecta |

---

# 10. F1-Score

## ¿Qué mide?
Combina precisión (Precision) y exhaustividad (Recall).

## Fórmula

\[
F1 =
2 \times
\frac{Precision \times Recall}
{Precision + Recall}
\]

## Interpretación
- Alto → Buen equilibrio entre precisión y recall.
- Bajo → Problemas en clasificación.

## Rango
| Valor | Significado |
|---|---|
| 0 | Peor rendimiento |
| 1 | Clasificación perfecta |

## Uso recomendado
Especialmente útil cuando:
- Las clases están desbalanceadas.
- Los falsos negativos son importantes.

```


# Métricas de Evaluación en Machine Learning

| Métrica | Tipo de Problema | ¿Qué responde? | Rango Ideal | ¿Cuándo priorizarla? |
|---|---|---|---|---|
| **Exactitud (Accuracy)** | Clasificación | ¿Qué porcentaje total de predicciones fueron correctas? | Cercano a 1 (100%) | Cuando las clases están balanceadas. Puede ser engañosa en datasets desbalanceados. |
| **Precisión (Precision)** | Clasificación | De todos los casos predichos como positivos, ¿cuántos realmente eran positivos? | Cercano a 1 | Cuando el costo de un Falso Positivo es alto (ej. detectar spam incorrectamente). |
| **Sensibilidad (Recall)** | Clasificación | De todos los positivos reales existentes, ¿cuántos detectó el modelo? | Cercano a 1 | Cuando el costo de un Falso Negativo es crítico (ej. diagnóstico médico). |
| **F1-Score** | Clasificación | ¿Cuál es el equilibrio entre Precision y Recall? | Cercano a 1 | En datasets desbalanceados donde importan tanto Falsos Positivos como Falsos Negativos. |
| **AUC-ROC** | Clasificación | ¿Qué tan bien separa el modelo las clases positivas y negativas independientemente del umbral? | 0.5 a 1.0 | Para comparar el desempeño general entre modelos de clasificación. |
| **R Cuadrado (R²)** | Regresión | ¿Qué proporción de la variabilidad del dato es explicada por el modelo? | Cercano a 1 | Para medir la capacidad explicativa global del modelo de regresión. |
| **MAE (Mean Absolute Error)** | Regresión | ¿Cuál es el error promedio absoluto de las predicciones? | Cercano a 0 | Cuando se desea interpretabilidad simple y los outliers no son críticos. |
| **RMSE (Root Mean Squared Error)** | Regresión | ¿Cuál es el error promedio penalizando más los errores grandes? | Cercano a 0 | Cuando los errores grandes deben castigarse fuertemente. |

---

# Interpretación General

## Clasificación

- **Accuracy** → rendimiento general.
- **Precision** → minimiza falsos positivos.
- **Recall** → minimiza falsos negativos.
- **F1-Score** → equilibrio entre Precision y Recall.
- **AUC-ROC** → capacidad discriminativa global del modelo.

## Regresión

- **R²** → calidad del ajuste.
- **MAE** → error promedio interpretable.
- **RMSE** → penaliza errores grandes.

---

# Ejemplo Práctico

## Diagnóstico Médico

- Si una enfermedad es grave:
  - Priorizar **Recall** para no dejar enfermos sin detectar.
- Si el tratamiento es riesgoso o caro:
  - Priorizar **Precision** para evitar falsos positivos.

## Predicción de Precios

- **MAE** → útil para conocer el error promedio real.
- **RMSE** → útil si grandes errores financieros son peligrosos.


# Rangos e Interpretación de Métricas en Machine Learning

## Métricas de Clasificación

| Métrica | Valor Mínimo | Significado del Mínimo | Valor Máximo | Significado del Máximo |
|---|---|---|---|---|
| **Accuracy (Exactitud)** | 0 | Ninguna predicción fue correcta. | 1 | Todas las predicciones fueron correctas. |
| **Precision** | 0 | Todos los positivos predichos fueron incorrectos (muchos falsos positivos). | 1 | Todos los positivos predichos eran realmente positivos. |
| **Recall (Sensibilidad)** | 0 | El modelo no detectó ningún positivo real. | 1 | El modelo detectó todos los positivos reales. |
| **F1-Score** | 0 | No existe equilibrio entre Precision y Recall; desempeño muy pobre. | 1 | Precision y Recall son perfectos. |
| **AUC-ROC** | 0 | El modelo clasifica completamente al revés de lo esperado. | 1 | Separación perfecta entre clases positivas y negativas. |

---

# Interpretación de AUC-ROC

| Rango | Interpretación |
|---|---|
| 0.5 | Modelo equivalente al azar |
| 0.6 – 0.7 | Rendimiento pobre |
| 0.7 – 0.8 | Rendimiento aceptable |
| 0.8 – 0.9 | Buen modelo |
| 0.9 – 1.0 | Excelente discriminación |

---

# Métricas de Regresión

| Métrica | Valor Mínimo | Significado del Mínimo | Valor Máximo | Significado del Máximo |
|---|---|---|---|---|
| **R² (Coeficiente de Determinación)** | Puede ser negativo | El modelo es peor que predecir usando la media. | 1 | El modelo explica perfectamente la variabilidad de los datos. |
| **MAE** | 0 | No existe error promedio en las predicciones. | ∞ | Error promedio extremadamente alto. |
| **RMSE** | 0 | No existe error cuadrático en las predicciones. | ∞ | Existen errores muy grandes en las predicciones. |

---

# Interpretación General de R²

| Rango | Interpretación |
|---|---|
| Menor que 0 | Modelo muy deficiente |
| 0 – 0.5 | Baja capacidad explicativa |
| 0.5 – 0.7 | Ajuste moderado |
| 0.7 – 0.9 | Buen ajuste |
| 0.9 – 1.0 | Excelente ajuste |

---

# Interpretación General de MAE y RMSE

## MAE

- Representa el error promedio absoluto.
- Se expresa en las mismas unidades del problema.
- Más fácil de interpretar.

### Ejemplo
Si el MAE = 2:
- En promedio, el modelo se equivoca en 2 unidades.

---

## RMSE

- Penaliza más los errores grandes.
- Sensible a outliers.
- Muy utilizado cuando grandes errores son peligrosos.

### Ejemplo
Si el RMSE = 10:
- Existen errores grandes que están aumentando considerablemente el promedio cuadrático.

---

# Ejemplo Práctico

## Clasificación Médica

| Métrica Prioritaria | Motivo |
|---|---|
| Recall | No dejar pacientes enfermos sin detectar. |
| Precision | Evitar diagnósticos falsos positivos. |
| F1-Score | Balance entre Recall y Precision. |

---

## Predicción de Ventas

| Métrica Prioritaria | Motivo |
|---|---|
| MAE | Error promedio fácil de interpretar. |
| RMSE | Penalizar errores financieros grandes. |
| R² | Medir qué tan bien el modelo explica las ventas. |
