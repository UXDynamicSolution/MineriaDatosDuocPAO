# Ejemplo de Clasificación y Regresión con Varios Modelos
Este notebook muestra ejemplos básicos con diferentes modelos de clasificación y regresión usando scikit-learn.

**Clasificación:**
- Árbol de Decisión
- K-Nearest Neighbors (KNN)
- Regresión Logística

**Regresión:**
- Regresión Lineal
- Árbol de Regresión
- K-Nearest Neighbors (Regressor)

```python
# Importaciones comunes
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.neighbors import KNeighborsClassifier, KNeighborsRegressor
from sklearn.metrics import accuracy_score, r2_score
```

## Clasificación: ¿Aprueba o no aprueba?

```python
# Datos
datos = pd.DataFrame({
    'horas_estudio': [1, 2, 3, 4, 5, 6, 7, 8],
    'asistencias':   [2, 3, 4, 4, 5, 6, 6, 7],
    'aprueba':       [0, 0, 0, 0, 1, 1, 1, 1]
})
X = datos[['horas_estudio', 'asistencias']]
y = datos['aprueba']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)
```

```python
# Árbol de Decisión
clf_tree = DecisionTreeClassifier().fit(X_train, y_train)
print('Árbol de decisión Accuracy:', accuracy_score(y_test, clf_tree.predict(X_test)))
```

```python
# KNN Clasificador
clf_knn = KNeighborsClassifier(n_neighbors=3).fit(X_train, y_train)
print('KNN Accuracy:', accuracy_score(y_test, clf_knn.predict(X_test)))
```

```python
# Regresión logística
clf_log = LogisticRegression().fit(X_train, y_train)
print('Regresión Logística Accuracy:', accuracy_score(y_test, clf_log.predict(X_test)))
```

## Regresión: Precio según m²

```python
# Datos
datos_reg = pd.DataFrame({
    'm2': [50, 60, 70, 80, 90, 100],
    'precio': [100, 120, 140, 160, 180, 200]
})
Xr = datos_reg[['m2']]
yr = datos_reg['precio']
Xr_train, Xr_test, yr_train, yr_test = train_test_split(Xr, yr, test_size=0.3)
```

```python
# Regresión lineal
reg_lin = LinearRegression().fit(Xr_train, yr_train)
print('Regresión lineal R²:', r2_score(yr_test, reg_lin.predict(Xr_test)))
```

```python
# Árbol de regresión
reg_tree = DecisionTreeRegressor().fit(Xr_train, yr_train)
print('Árbol de regresión R²:', r2_score(yr_test, reg_tree.predict(Xr_test)))
```

```python
# KNN Regresor
reg_knn = KNeighborsRegressor(n_neighbors=3).fit(Xr_train, yr_train)
print('KNN Regresor R²:', r2_score(yr_test, reg_knn.predict(Xr_test)))
```
